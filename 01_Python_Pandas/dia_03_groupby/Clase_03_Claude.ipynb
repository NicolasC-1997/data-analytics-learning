```python
# ============================================================
# DÍA 3 - Agrupación y Agregación con .groupby()
# Programa Data Analytics - 90 días
# ============================================================

import pandas as pd

# Cargar el dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Dataset cargado correctamente")
print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")
```

```python
# ============================================================
# BLOQUE 1: groupby básico - una columna de agrupación
# ============================================================

# Tasa de supervivencia por clase
supervivencia_por_clase = df.groupby('Pclass')['Survived'].mean()
print("=== Tasa de supervivencia por clase ===")
print(supervivencia_por_clase)
print()

# Interpretación: Survived es 0 o 1, así que la media = porcentaje de supervivencia
# Esto es un patrón muy común en análisis de datos binarios
```

```python
# ============================================================
# BLOQUE 2: Múltiples funciones con .agg()
# ============================================================

# Análisis de edad por clase
edad_por_clase = df.groupby('Pclass')['Age'].agg(['mean', 'median', 'min', 'max', 'count'])
print("=== Análisis de edad por clase ===")
print(edad_por_clase)
print()

# .agg() permite obtener varias métricas en una sola operación
# En la industria esto se usa constantemente para hacer perfiles de grupos
```

```python
# ============================================================
# BLOQUE 3: Agrupar por dos columnas simultáneamente
# ============================================================

# Supervivencia por clase Y sexo
supervivencia_clase_sexo = df.groupby(['Pclass', 'Sex'])['Survived'].mean()
print("=== Supervivencia por clase y sexo ===")
print(supervivencia_clase_sexo)
print()

# Nota: cuando agrupamos por múltiples columnas obtenemos un MultiIndex
# Veremos cómo aplanarlo más adelante
```

```python
# ============================================================
# BLOQUE 4: .value_counts() - contar categorías
# ============================================================

# Cuántos pasajeros hay en cada clase
pasajeros_por_clase = df['Pclass'].value_counts()
print("=== Pasajeros por clase ===")
print(pasajeros_por_clase)
print()

# value_counts() es un atajo muy útil para contar valores únicos
# Equivale a groupby + count, pero más directo para una sola columna
```

```python
# ============================================================
# BLOQUE 5: reset_index() - convertir el resultado en DataFrame
# ============================================================

# Por defecto, groupby devuelve una Serie con el índice como la columna de agrupación
# reset_index() lo convierte en un DataFrame normal, más fácil de trabajar

resultado = df.groupby('Pclass')['Survived'].mean().reset_index()
resultado.columns = ['Clase', 'Tasa_Supervivencia']
resultado['Tasa_Supervivencia_Pct'] = (resultado['Tasa_Supervivencia'] * 100).round(1)

print("=== Resultado como DataFrame ===")
print(resultado)
```

```python
# ============================================================
# BLOQUE 6: Ejemplo de negocio completo
# ============================================================

# Pregunta de negocio: "¿Qué perfil tienen los sobrevivientes vs. no sobrevivientes?"
# Esta es exactamente la pregunta que haría un analista en una reunión

perfil = df.groupby('Survived').agg(
    cantidad=('PassengerId', 'count'),
    edad_promedio=('Age', 'mean'),
    tarifa_promedio=('Fare', 'mean'),
    proporcion_mujeres=('Sex', lambda x: (x == 'female').mean())
).round(2)

perfil.index = ['No sobrevivió', 'Sobrevivió']
print("=== Perfil de sobrevivientes vs. no sobrevivientes ===")
print(perfil)
```
