```python
# ============================================================
# DÍA 1 - Exploración Básica con Pandas
# Dataset: Titanic
# ============================================================

import pandas as pd
import numpy as np

# ── Carga de datos ──────────────────────────────────────────
# Cargamos directamente desde la URL pública de Seaborn/GitHub
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("=" * 50)
print("PASO 1: ¿Qué hay en este dataset?")
print("=" * 50)

# ── Exploración inicial ─────────────────────────────────────
print(f"\nDimensiones: {df.shape[0]} filas × {df.shape[1]} columnas\n")
print("Primeras 5 filas:")
print(df.head())

print("\n" + "=" * 50)
print("PASO 2: Información de estructura")
print("=" * 50)
df.info()

print("\n" + "=" * 50)
print("PASO 3: Estadísticas descriptivas")
print("=" * 50)
print(df.describe())

print("\n" + "=" * 50)
print("PASO 4: Valores nulos por columna")
print("=" * 50)
nulos = df.isnull().sum()
porcentaje = (df.isnull().sum() / len(df) * 100).round(2)
resumen_nulos = pd.DataFrame({
    "Nulos": nulos,
    "Porcentaje (%)": porcentaje
})
print(resumen_nulos[resumen_nulos["Nulos"] > 0])  # Solo columnas con nulos

print("\n" + "=" * 50)
print("PASO 5: Tipos de datos")
print("=" * 50)
print(df.dtypes)

print("\n" + "=" * 50)
print("PASO 6: Columnas disponibles")
print("=" * 50)
print(list(df.columns))
```
