```python
# ============================================================
# DÍA 2 - Selección y Filtrado con Pandas
# Dataset: Titanic
# ============================================================

import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("=" * 30)
print("Columnas disponibles" + "\n")
print(list(df.columns))
print("=" * 30)
# ─────────────────────────────────────────────────────────────
# PREGUNTA 1
# ¿Cuántas mujeres había a bordo del Titanic?
# ─────────────────────────────────────────────────────────────
# Tu código aquí:
mujeres = df[(df['Sex'] == 'female')]
print("Mujeres a bordo: ", len(mujeres))
print("=" * 30 + "\n")
# ─────────────────────────────────────────────────────────────
# PREGUNTA 2
# ¿Cuál es la edad promedio de los pasajeros de primera clase?
# ─────────────────────────────────────────────────────────────
# Tu código aquí:
primera_clase = df[df['Pclass'] == 1]
edad_promedio = primera_clase['Age'].mean()
print(f"Edad promedio, 1ra clase: {edad_promedio:.1f} años")
print("=" * 30 + "\n")
# ─────────────────────────────────────────────────────────────
# PREGUNTA 3
# ¿Cuántos menores de 18 años sobrevivieron?
# ─────────────────────────────────────────────────────────────
# Tu código aquí:

menores_sobrevivientes = df[(df['Survived'] == 1) & (df['Age'] < 18)]
print(f"Menores que sobrevivieron :  {len(menores_sobrevivientes)}")
print("=" * 30 + "\n")
# ─────────────────────────────────────────────────────────────
# PREGUNTA 4
# Muestra el nombre, edad y tarifa de los 5 pasajeros
# de primera clase que pagaron más caro.
# ─────────────────────────────────────────────────────────────
# Tu código aquí:

primera_clase = df[df['Pclass'] == 1]
top5 = primera_clase[['Name', 'Age', 'Fare']].sort_values('Fare', ascending=False).head(5)
print("Los 5 boletos mas caros")
print(top5.to_string(index=False))
print("=" * 30 + "\n")
# ─────────────────────────────────────────────────────────────
# PREGUNTA 5 (desafío)
# ¿Qué porcentaje de mujeres en primera clase sobrevivió?
# Pista: combina filtros y usa .mean() sobre la columna Survived
# ─────────────────────────────────────────────────────────────
# Tu código aquí:

mujeres_primera = df[(df['Pclass'] == 1) & (df['Sex'] == 'female')]
tasa_supervivencia = mujeres_primera['Survived'].mean() * 100
print(f"Tasa de supervivencia, mujeres de primera clase: {tasa_supervivencia:.1f}%")
```
